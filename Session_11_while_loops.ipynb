```python
for i in range(1,11):
    print(i,end=' ')

for i in range(10,0,-1):
    print(i,end=' ')

for i in range(-10,0):
    print(i,end=' ')

for i in range(-1,-11,-1):
    print(i,end=' ')
```

```python
i=1
while i<11:
    print(i,end=' ')
    i=i+1
```

```python
i=10  #mandatory
while i>0: #mandatory
    print(i,end=' ')
    i=i-1  #mandatory
```

```python
i=1
while True:  # suppose we write only True
    print(i,end=' ')
    i=i+1
    if i==11:  # we can apply condition here
        break  # and make the loop stop by breaking.
```
