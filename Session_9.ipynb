```python
def number(n): #n is defined when we are calling the function
    if n==0:   #condition is that if n==0 then return 0
        return(0)
    print(n)   # print dediya
    number(n-1) # now number is changed.
         
number(6) 

#summary step by step is:-
#defined the function.
#if condition rokne ki liye
#fir print karwa diya
#number(ek kam karwa diya Recursion k Lye)
```

```python
def factorial(n):
    if n==0 or n==1: #base case stops at 0 or 1
        return(1)
    return(n*factorial(n-1))
print(factorial(5))
```

```python
def fact(n): #n is defined when we are calling the function
    if n==0:   #condition is that if n==0 then return 1 otherwise sub kuch zero ho jayega
        return(1)
    else:
        return(n*fact(n-1))
             
fact(5) 

#summary step by step is:-
#defined the function.
#if condition rokne ki liye 
#6==0 ----false---- print(6*fact(5))
#5==0 ----false---- print(5*fact(4))
#4==0 ----false---- print(4*fact(3))
#3==0 ----false---- print(3*fact(2))
#2==0 ----false---- print(2*fact(1))
#1==0 ----false---- print(1*fact(0))
#0==0 ----true---- print(1*fact(0))===1
#number(ek kam karwa diya Recursion k Lye)
```
