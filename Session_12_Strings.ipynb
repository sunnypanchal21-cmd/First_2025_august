```python
emails=['virat.kholi@rcb.com','rohit.sharma@mi.com','ms.dhoni@csk.com']



num=eval(input('enter the string number(0,1,2):'))
email=emails[num]

f_dot=email.index('.')
s_dot=email.index('@')
t_dot=email.index('.',f_dot+1)

first_name=email[:f_dot]
second_name=email[f_dot+1:s_dot]
company_name=email[s_dot+1:t_dot]

print('first_name:',first_name)
print('second_name:',second_name)
print('company_name:', company_name)
```
