```python
#WAP the user get the random number between 1 to 100
#Find Even and odd.
#get sum_of_Even and sum_of_odd
#get 5 random number

import random
sum_of_even=0
sum_of_odd=0
for i in range(1,6):
    num=random.randint(1,100)
    if num%2==0:
        print(f"The {num} is even")
        sum_of_even=sum_of_even+num
        
    else:
        print(f"The {num} is odd")
        sum_of_odd+=num 

print("sum_of_even:-", sum_of_even)
print("sum_of_odd:-", sum_of_odd)
```
