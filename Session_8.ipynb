```python
num1=eval(input("Enter number 1:-"))
num2=eval(input("Enter number 2:-"))
print("Enter  + \n  - \n  * \n  /")
operation=input("enter the symbol [+,-,*,/]")
if operation=="+":
    print("Addition operation is selected")
    PO=num1+num2
    print(f"The addition of {num1} and {num2} is {PO}")
elif operation=="-":
    print("subtraction operation is selected")
    PO=num1-num2
    print(f"The subtraction of {num1} and {num2} is {PO}")
elif operation=="*":
    print("multiplication operation is selected")
    PO=num1*num2
    print(f"The multiplication of {num1} and {num2} is {PO}")
elif operation=="/":
    print("Division operation is selected")
    PO=num1/num2
    print(f"The division of {num1} and {num2} is {PO}")
```
